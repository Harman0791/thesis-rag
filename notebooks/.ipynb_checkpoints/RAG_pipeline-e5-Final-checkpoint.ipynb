```python
import re, unicodedata, string,json
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

def extract_answer_text(ans):
    if isinstance(ans, dict):
        val = ans.get("value", "")
        aliases = ans.get("aliases", [])
        nor_val = ans.get("normalized_value", "")
        nor_aliases = ans.get("normalized_aliases", "")
        all_vals = [val] + [a for a in aliases if a and a != val]
        return " , ".join(sorted(set(map(str, all_vals))))
    elif isinstance(ans, list):
        texts = []
        for item in ans:
            if isinstance(item, dict):
                texts.append(item.get("value", ""))
                texts.extend(item.get("aliases", []))
                texts.append(item.get("normalized_value", ""))
                texts.append(item.get("normalized_aliases", ""))
        return " , ".join(sorted(set(filter(None, texts))))
    else:
        return str(ans) if ans else ""
    
def extract_wiki_context(entitypage):
    import numpy as np
    
    def clean_text(text):
        """Remove HTML tags, brackets, punctuation, and special characters."""
        if not isinstance(text, str):
            text = str(text)
        # Remove HTML tags (if any)
        text = re.sub(r"<.*?>", " ", text)
        # Remove brackets and their contents like [text], (text), {text}
        text = re.sub(r"\[.*?\]|\(.*?\)|\{.*?\}", " ", text)
        # Remove punctuation and special characters (keep letters, digits, and spaces)
        text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
        # Replace multiple spaces with a single space
        text = re.sub(r"\s+", " ", text)
        return text.strip()

    def words_from(x):
        """Flatten strings/lists/ndarrays (even nested) into a list of words."""
        if x is None:
            return []
        if isinstance(x, np.ndarray):
            return words_from(x.tolist())
        if isinstance(x, (list, tuple, set)):
            out = []
            for e in x:
                out.extend(words_from(e))
            return out
        return clean_text(str(x)).split()

    def content(x):
        w = words_from(x)
        return " ".join(w)

    # Case 1: a single entity page dict with arrays for each field
    if isinstance(entitypage, dict):
        return content(entitypage.get("wiki_context", []))

    # Case 2: a list of such dicts -> merge all wiki_contexts, then truncate
    if isinstance(entitypage, list):
        all_words = []
        for item in entitypage:
            if isinstance(item, dict):
                all_words += words_from(item.get("wiki_context", []))
            else:
                all_words += words_from(item)
        return " ".join(all_words)

    # Fallback
    return content(entitypage)

assert "question" in df.columns, "df must have a 'question' column"
if "question_id" in df.columns:
    df["question_id"] = df["question_id"].astype("string")
else:
    # Fall back to index, but keep it as a string ID
    df["question_id"] = pd.Series(df.index, index=df.index).astype("Int64").astype("string")

df["question_clean"] = df["question"].astype(str).str.strip()
df["answer_text"]    = df["answer"].map(extract_answer_text)
df["ctx"] = df["entity_pages"].map(extract_wiki_context)


df_q_eval = df[["question_id", "question_clean", "answer_text"]].copy()

df_q_eval = df_q_eval.drop_duplicates(subset=["question_id"]).reset_index(drop=True)

df_q_eval.to_parquet(Q_PATH, index=False)
print("Column Name :", df.columns)
print(type(df["ctx"]))
print(df["ctx"].dtype)
print(df["ctx"].head(2))
pd.set_option('display.max_colwidth', None)
print(df_q_eval.head(1))
```

```python
       
def _to_parquet_safe(df_out: pd.DataFrame, path: str | Path):
    try:
        df_out.to_parquet(path, engine="pyarrow", index=False)
    except Exception:
        try:
            df_out.to_parquet(path, engine="fastparquet", index=False)
        except Exception as e:
            raise RuntimeError(
                "Saving to Parquet failed. Please install 'pyarrow' or 'fastparquet'. "
                f"Original error: {e}"
            )
```

```python
import pandas as pd
import numpy as np

def build_passages_token_based(
    df,
    context_cols,
    doc_id_col="question_id",
    chunk_size=128,
    overlap_ratio=0.1,
    dedupe=True,
    keep_source=False
):
    """
    Robust passage builder:
    - If df already looks like a passages table (has 'text'), it will use that automatically
      when requested context_cols are missing.
    - Uses tokenizer overflow windows (<=512 tokens) to avoid long-sequence warnings.
    """

    context_cols = list(context_cols)

    # --- If the requested context columns are missing, auto-fallback to 'text' ---
    missing_ctx = [c for c in context_cols if c not in df.columns]
    if missing_ctx:
        if "text" in df.columns:
            # df is already a passages dataframe
            context_cols = ["text"]
        else:
            raise KeyError(
                f"context_cols missing from df.columns: {missing_ctx}. "
                f"Available: {list(df.columns)[:30]}..."
            )

    # --- doc_id handling: fallback to 'doc_id' if question_id isn't present ---
    if doc_id_col not in df.columns and "doc_id" in df.columns:
        doc_id_col = "doc_id"

    has_doc_id = (doc_id_col in df.columns)

    # --- Determine max len (E5 is 512) ---
    MAX_LEN = getattr(tokenizer, "model_max_length", 512)
    if MAX_LEN is None or MAX_LEN > 100000:
        MAX_LEN = 512
    MAX_LEN = min(int(MAX_LEN), 512)

    # Respect caller chunk_size but never exceed model limit
    chunk_size = min(int(chunk_size), MAX_LEN)

    # Stride = overlap in tokens
    stride = int(chunk_size * overlap_ratio)
    if stride < 0:
        stride = 0
    if stride >= chunk_size:
        stride = max(0, chunk_size - 1)

    rows = []

    # Iterate only on columns that exist
    cols = ([doc_id_col] if has_doc_id else []) + context_cols

    for r in df[cols].itertuples(index=True, name=None):
        idx = r[0]

        if has_doc_id:
            doc_id_val = r[1]
            doc_id = doc_id_val if pd.notna(doc_id_val) else idx
            ctx_values = r[2:]
        else:
            doc_id = idx
            ctx_values = r[1:]

        parts = [str(x) for x in ctx_values if pd.notna(x)]
        text = "\n".join(parts).strip()
        if not text:
            continue

        enc = tokenizer(
            text,
            add_special_tokens=False,
            return_offsets_mapping=True,
            truncation=True,
            max_length=chunk_size,
            return_overflowing_tokens=True,
            stride=stride
        )

        offset_windows = enc.get("offset_mapping", [])
        if not offset_windows:
            continue

        # Safety cap: number of chunks per doc
        MAX_CHUNKS_PER_DOC = 20000
        if len(offset_windows) > MAX_CHUNKS_PER_DOC:
            offset_windows = offset_windows[:MAX_CHUNKS_PER_DOC]

        for i, offsets in enumerate(offset_windows):
            if not offsets:
                continue

            char_start = offsets[0][0]
            char_end = offsets[-1][1]
            chunk_text = text[char_start:char_end].strip()
            if not chunk_text:
                continue

            passage_id = f"{doc_id}::r{idx}::c{i}"

            rec = {"passage_id": passage_id, "doc_id": doc_id, "text": chunk_text}
            if keep_source:
                rec["source"] = context_cols[0] if context_cols else None
            rows.append(rec)

    passages = pd.DataFrame(rows)
    if dedupe and not passages.empty:
        passages = passages.drop_duplicates(subset=["text"]).reset_index(drop=True)

    return passages
```

```python
# --- BUILD & SAVE PASSAGES (token-based chunking) ---
from pathlib import Path
from sentence_transformers import SentenceTransformer

MODEL_NAME = "intfloat/e5-small-v2"
model = SentenceTransformer(MODEL_NAME)
model.max_seq_length = 512
tokenizer = model.tokenizer

# Build passages *only from contexts*
passages = build_passages_token_based(
    df=df,
    context_cols=[ctx_col_name],   # e.g. ["context_clean"] also works if that's your column
    doc_id_col="question_id",
    chunk_size=min(CHUNK_SIZE, model.max_seq_length),
    overlap_ratio=OVERLAP_RATIO,
    dedupe=True,
    keep_source=False
)

dups = passages["passage_id"].duplicated().sum()
print("Duplicate passage_id count:", dups)
assert dups == 0, "passage_id must be unique for retrieval mapping"


_to_parquet_safe(passages, PASSAGES_PATH)

n_docs = df.shape[0]
n_pass = len(passages)
avg_chunks = (n_pass / n_docs) if n_docs else 0

print(f"Saved {n_pass:,} passages to '{PASSAGES_PATH}'. "
      f"(chunk_size={CHUNK_SIZE}, overlap_ratio={OVERLAP_RATIO})")
print(f"Docs: {n_docs:,} | Avg chunks/doc: {avg_chunks:.2f}")

passages.head(3)
```

## Embedding Script

```python
import time
import math, os
import psutil
import numpy as np
import pandas as pd
from tqdm import tqdm
from sentence_transformers import SentenceTransformer, models

# ------------------ Config ------------------
MODEL_NAME   = "intfloat/e5-small-v2"  # 384-dim, fast on CPU
BATCH_SIZE   = 128                  # choose 64–128 for CPU
EMB_DTYPE    = "float32"            # "float32" or "float16"
PASSAGES_DF  = passages             # your DataFrame with passage_id, doc_id, text

MODEL_DIRNAME = MODEL_NAME.replace("/", "_")
EMB_DIR  = PROJECT_ROOT / "embeddings" / MODEL_DIRNAME / CFG_NAME
EMB_DIR.mkdir(parents=True, exist_ok=True)

EMB_PATH     = EMB_DIR/"emb.npy"
IDS_PATH     = EMB_DIR/"ids.npy"

os.makedirs(os.path.dirname(EMB_PATH), exist_ok=True)
os.makedirs(os.path.dirname(IDS_PATH), exist_ok=True)
# -------------------------------------------

# ---- Validate input DataFrame ----
required_cols = {"passage_id", "doc_id", "text"}
missing = required_cols - set(PASSAGES_DF.columns)
if missing:
    raise KeyError(f"passages DataFrame missing required columns: {missing}")

# Drop rows with empty/NaN chunk_text (keep order stable)
df = PASSAGES_DF.copy()
df["text"] = df["text"].astype(str).str.strip()
df = df[df["text"] != ""].sort_values("passage_id").reset_index(drop=True)
if df.empty:
    raise ValueError("No non-empty passages to embed.")


texts = [f"passage: {t}" for t in df["text"].tolist()]
df["passage_id"] = df["passage_id"].astype("string")

# --- canonical passages used for embedding/indexing ---
df = df.sort_values("passage_id").reset_index(drop=True)

df["faiss_id"] = np.arange(len(df), dtype=np.int64)

CANONICAL_PASSAGES_PATH = EMB_DIR / "passages_canonical.parquet"
ID_MAP_PATH = EMB_DIR / "id_map.csv"

df.to_parquet(CANONICAL_PASSAGES_PATH, index=False)
df[["faiss_id", "passage_id"]].to_csv(ID_MAP_PATH, index=False)

# ids used for add_with_ids
ids = df["faiss_id"].values.astype(np.int64)

print("Saved canonical passages:", CANONICAL_PASSAGES_PATH)
print("Saved id_map:", ID_MAP_PATH)

```

```python
from pathlib import Path
import os, time, json
import numpy as np
from sentence_transformers import SentenceTransformer

os.environ["TOKENIZERS_PARALLELISM"] = "false"

# Load model
model = SentenceTransformer(MODEL_NAME)

# Paths (use the same ones defined earlier)
EMB_PATH  = Path(EMB_DIR / "emb.npy")
IDS_PATH  = Path(EMB_DIR / "ids.npy")
META_PATH = Path(EMB_DIR / "meta.json")

# IMPORTANT: build texts from the canonical df so ids/texts align 1:1
texts = [f"passage: {t}" for t in df["text"].astype(str).tolist()]
ids   = df["faiss_id"].values.astype(np.int64)

N = len(texts)
assert N == len(ids), f"Mismatch: N texts {N} vs ids {len(ids)}"

D = model.get_sentence_embedding_dimension()
dtype = np.dtype(EMB_DTYPE)

if EMB_PATH.exists() and IDS_PATH.exists() and META_PATH.exists():
    print("Embeddings already exist — skipping embedding step.")
    elapsed = 0.0  # so later prints don't crash
else:
    # Remove stale files (Windows-friendly)
    for p in [EMB_PATH, IDS_PATH, META_PATH]:
        try:
            if p.exists():
                p.unlink()
        except PermissionError:
            raise RuntimeError(f"{p} appears to be in use. Close readers and retry.")

    # Create memmap
    Xm = np.lib.format.open_memmap(
        str(EMB_PATH),
        mode="w+",
        dtype=dtype,
        shape=(N, D),
    )

    print(f"Memmap ready at {EMB_PATH} with shape {Xm.shape} and dtype {Xm.dtype}.")

    t0 = time.time()
    w = 0

    for i in range(0, N, BATCH_SIZE):
        batch = texts[i:i + BATCH_SIZE]
        vecs = model.encode(
            batch,
            batch_size=BATCH_SIZE,
            convert_to_numpy=True,
            normalize_embeddings=True,
            show_progress_bar=False,
        ).astype(dtype, copy=False)

        n = vecs.shape[0]
        Xm[w:w+n, :] = vecs
        w += n

    elapsed = time.time() - t0
    assert w == N, f"Wrote {w} vectors but expected {N}"

    # Flush + close memmap (important on Windows)
    Xm.flush()
    del Xm

    # Save ids + metadata
    np.save(IDS_PATH, ids)

    with open(META_PATH, "w", encoding="utf-8") as f:
        json.dump(
            {
                "model": MODEL_NAME,
                "dim": int(D),
                "n_vecs": int(N),
                "dtype": str(dtype),
                "normalize": True,
                "batch_size": int(BATCH_SIZE),
                "time_sec": round(elapsed, 2),
                "throughput_passages_per_sec": round(N / elapsed, 1) if elapsed > 0 else None,
            },
            f,
            indent=2,
        )

print(f"Embeddings: {(N, D)} saved → {EMB_PATH}")
print(f"IDs saved → {IDS_PATH}")
print(f"Meta saved → {META_PATH}")
print(f"Time: {elapsed:.2f}s | Throughput: {(N/elapsed):.1f}/s" if elapsed > 0 else "Time: skipped")
```

## Build FAISS HNSW Index

```python
# --- HNSW index builder: FAISS IndexHNSWFlat (cosine/IP) ---
import os, json, time
from pathlib import Path

import numpy as np
import psutil

try:
    import faiss  # CPU FAISS
except ImportError as e:
    raise RuntimeError("faiss is required. Install with: pip install faiss-cpu") from e

# ------------------ CONFIG ------------------
PROJECT_ROOT = Path(globals().get("PROJECT_ROOT", Path.cwd()))
MODEL_NAME   = globals().get("MODEL_NAME", "intfloat/e5-small-v2")

# Use the same EMB_DIR from your embedding step
EMB_PATH   = Path(EMB_DIR / "emb.npy")
IDS_PATH   = Path(EMB_DIR / "ids.npy")
META_PATH  = Path(EMB_DIR / "meta.json")

# HNSW hyperparameters
HNSW_M             = int(os.environ.get("HNSW_M", 32))
HNSW_EF_CONSTRUCT  = int(os.environ.get("HNSW_EF_CONSTRUCT", 200))
HNSW_EF_SEARCH     = int(os.environ.get("HNSW_EF_SEARCH", 64))

# Add in batches
ADD_BATCH_SIZE     = int(os.environ.get("HNSW_ADD_BATCH_SIZE", 100_000))

# Save index in a consistent directory (avoid mixing runs)
MODEL_DIRNAME = MODEL_NAME.replace("/", "_")
INDEX_DIR = PROJECT_ROOT / "indexes" / MODEL_DIRNAME / CFG_NAME / f"hnsw_M{HNSW_M}_efC{HNSW_EF_CONSTRUCT}_efS{HNSW_EF_SEARCH}"
INDEX_DIR.mkdir(parents=True, exist_ok=True)
INDEX_PATH = INDEX_DIR / "index.faiss"
INDEX_META = INDEX_DIR / "meta.json"

# Limit FAISS threads (optional)
try:
    n_threads = max(1, min(os.cpu_count() or 1, 8))
    faiss.omp_set_num_threads(n_threads)
except Exception:
    n_threads = None

# ------------------ LOAD EMBEDDINGS (MMAP) ------------------
if not EMB_PATH.exists():
    raise FileNotFoundError(f"Embeddings file not found: {EMB_PATH}")
if not IDS_PATH.exists():
    raise FileNotFoundError(f"IDs file not found: {IDS_PATH}")

# Read meta if present
if META_PATH.exists():
    with open(META_PATH, "r", encoding="utf-8") as f:
        emb_meta = json.load(f)
    D = int(emb_meta.get("dim"))
    N = int(emb_meta.get("n_vecs"))
else:
    X_tmp = np.load(EMB_PATH, mmap_mode="r")
    if X_tmp.ndim != 2:
        raise ValueError(f"Expected (N,D) embeddings array; got shape {X_tmp.shape}")
    N, D = int(X_tmp.shape[0]), int(X_tmp.shape[1])

# Memory-map embeddings and ids
X = np.load(EMB_PATH, mmap_mode="r")
ids = np.load(IDS_PATH, mmap_mode="r")

if len(ids) != N:
    raise ValueError(f"IDs length {len(ids)} != embeddings N {N}")

need_cast = (X.dtype != np.float32)

# ------------------ BUILD INDEX (IP/COSINE) ------------------
# Normalized embeddings => cosine == inner product
try:
    base = faiss.IndexHNSWFlat(D, HNSW_M, faiss.METRIC_INNER_PRODUCT)
except TypeError:
    base = faiss.IndexHNSWFlat(D, HNSW_M)
    if hasattr(base, "metric_type"):
        base.metric_type = faiss.METRIC_INNER_PRODUCT

base.hnsw.efConstruction = HNSW_EF_CONSTRUCT
base.hnsw.efSearch = HNSW_EF_SEARCH

# Wrap to store your faiss_id values
index = faiss.IndexIDMap2(base)

# Remove stale artifact before writing
try:
    if INDEX_PATH.exists():
        INDEX_PATH.unlink()
except PermissionError:
    raise RuntimeError(f"{INDEX_PATH} is in use; close readers and retry.")

proc = psutil.Process(os.getpid())
def rss_mb() -> float:
    return proc.memory_info().rss / (1024 * 1024)

peak_rss_mb = rss_mb()

t0 = time.time()
added = 0

for start in range(0, N, ADD_BATCH_SIZE):
    end = min(start + ADD_BATCH_SIZE, N)
    sl = slice(start, end)

    # embeddings batch (float32, C-contiguous)
    xb = np.asarray(X[sl], dtype=np.float32, order="C") if need_cast else np.asarray(X[sl], dtype=np.float32, order="C")
    # ids batch (int64, C-contiguous)
    id_batch = np.asarray(ids[sl], dtype=np.int64, order="C")

    index.add_with_ids(xb, id_batch)

    added += (end - start)
    peak_rss_mb = max(peak_rss_mb, rss_mb())

elapsed = time.time() - t0
assert index.ntotal == added, f"Index holds {index.ntotal} but added {added}"

# ------------------ SAVE INDEX + STATS ------------------
faiss.write_index(index, str(INDEX_PATH))
file_size_mb = INDEX_PATH.stat().st_size / (1024 * 1024)

index_meta = {
    "type": "faiss.IndexHNSWFlat + IndexIDMap2",
    "model": MODEL_NAME,
    "embeddings_path": str(EMB_PATH),
    "ids_path": str(IDS_PATH),
    "n_vecs": int(N),
    "dim": int(D),
    "normalized": True,
    "metric": "inner_product",
    "hnsw": {"M": HNSW_M, "efConstruction": HNSW_EF_CONSTRUCT, "efSearch": HNSW_EF_SEARCH},
    "build": {
        "time_sec": round(elapsed, 2),
        "throughput_vecs_per_sec": round(N / elapsed, 1) if elapsed > 0 else None,
        "peak_rss_mb": round(peak_rss_mb, 1),
        "faiss_threads": n_threads,
    },
    "artifacts": {
        "index_path": str(INDEX_PATH),
        "index_disk_size_mb": round(file_size_mb, 1),
        "index_dir": str(INDEX_DIR),
        "meta_path": str(INDEX_META),
    },
    "created_utc": time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime()),
}

with open(INDEX_META, "w", encoding="utf-8") as f:
    json.dump(index_meta, f, indent=2)

print(f"[OK] HNSW index saved → {INDEX_PATH}")
print(f"     Vectors: {N} | Dim: {D} | Metric: IP(cosine)")
print(f"     HNSW(M={HNSW_M}, efC={HNSW_EF_CONSTRUCT}, efS={HNSW_EF_SEARCH})")
print(f"     Build time: {elapsed:.2f}s | Peak RSS: {peak_rss_mb:.1f} MB | File: {file_size_mb:.1f} MB")
print(f"     Meta: {INDEX_META}")
```

## Retrieval Step - HyBrid Retrieval

```python
from typing import Optional, Tuple, List, Dict
from pathlib import Path
import time
import numpy as np
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer

class DenseRetriever:
    """
    Dense retriever on FAISS HNSW index with timing hooks.
    - Encodes query with the SAME model as passages.
    - Normalizes query embeddings (cosine/IP match).
    - Timed search returns t_embed_q (ms), t_search (ms), t_total_retrieval (ms).
    """

    def __init__(
        self,
        model_name: str,
        index_path: Path,
        passages_df: pd.DataFrame,
        id_map_df: Optional[pd.DataFrame] = None,
        normalize_embeddings: bool = True,
        device: str = "cpu",
        debug: bool = False,
    ):
        self.model_name = model_name
        self.index_path = Path(index_path)
        self.normalize = normalize_embeddings
        self.device = device
        self.debug = debug

        if not self.index_path.exists():
            raise FileNotFoundError(f"Index not found: {self.index_path}")

        # Load FAISS index
        self.index = faiss.read_index(str(self.index_path))

        # Load embedding model
        self.model = SentenceTransformer(model_name, device=device)

        # Store passages (canonical df is recommended)
        self.passages_df = passages_df.copy()
        if "passage_id" not in self.passages_df.columns:
            raise KeyError("passages_df must contain 'passage_id'")
        self.passages_df["passage_id"] = self.passages_df["passage_id"].astype(str)

        # --- DEBUG: check uniqueness assumptions ---
        if self.debug:
            print("\n[DEBUG:init] passages_df rows:", len(self.passages_df))
            print("[DEBUG:init] passages_df passage_id unique?:", self.passages_df["passage_id"].is_unique)
            if "faiss_id" in self.passages_df.columns:
                print("[DEBUG:init] passages_df faiss_id unique?:", self.passages_df["faiss_id"].is_unique)

        # Build mapping from FAISS id -> passage_id
        self.int_to_pid = None
        if id_map_df is not None:
            cols = set(id_map_df.columns)
            if {"faiss_id", "passage_id"} <= cols:
                if self.debug:
                    print("\n[DEBUG:init] id_map_df rows:", len(id_map_df))
                    print("[DEBUG:init] id_map_df faiss_id unique?:", id_map_df["faiss_id"].is_unique)
                    print("[DEBUG:init] id_map_df passage_id unique?:", id_map_df["passage_id"].is_unique)

                self.int_to_pid = dict(
                    zip(
                        id_map_df["faiss_id"].astype(np.int64),
                        id_map_df["passage_id"].astype(str),
                    )
                )

                if self.debug:
                    print("[DEBUG:init] int_to_pid size:", len(self.int_to_pid))

            elif {"int_id", "passage_id"} <= cols:
                self.int_to_pid = dict(
                    zip(
                        id_map_df["int_id"].astype(np.int64),
                        id_map_df["passage_id"].astype(str),
                    )
                )

                if self.debug:
                    print("\n[DEBUG:init] (legacy) int_to_pid size:", len(self.int_to_pid))

    def set_ef_search(self, ef: int):
        """Set efSearch on underlying HNSW index (handles IndexIDMap2 wrapper)."""
        inner = self.index
        if isinstance(inner, (faiss.IndexIDMap, faiss.IndexIDMap2)):
            inner = inner.index
        if hasattr(inner, "hnsw"):
            inner.hnsw.efSearch = int(ef)

    def _encode_query(self, q: str) -> Tuple[np.ndarray, float]:
        t0 = time.perf_counter()
        q_emb = self.model.encode(
            [f"query: {q}"],
            convert_to_numpy=True,
            normalize_embeddings=self.normalize,
            show_progress_bar=False,
        ).astype("float32", copy=False)
        t_ms = (time.perf_counter() - t0) * 1000.0
        return q_emb, t_ms

    def _ids_to_rows(self, ids: List[int]) -> pd.DataFrame:
        """
        Convert FAISS integer ids into rows of passages_df.
        Preserves order and length (left-join), so downstream alignment stays safe.
        """
        if self.int_to_pid is not None:
            pids = [self.int_to_pid.get(int(i)) for i in ids]

            if self.debug:
                # how many mapped vs missing?
                n_none = sum(x is None for x in pids)
                print("\n[DEBUG:_ids_to_rows] len(ids):", len(ids))
                print("[DEBUG:_ids_to_rows] len(pids):", len(pids), "missing passage_id:", n_none)
                # duplicates in pids can cause merge expansion if passages_df has duplicate passage_id
                print("[DEBUG:_ids_to_rows] unique pids (incl None):", len(set(pids)))

            ordered = pd.DataFrame({"passage_id": pd.Series(pids, dtype="string")})

            # DEBUG: check if passages_df has duplicate passage_id
            if self.debug:
                dup_pid = self.passages_df["passage_id"].duplicated().sum()
                print("[DEBUG:_ids_to_rows] passages_df duplicate passage_id count:", int(dup_pid))

            if self.debug:
                print("[DEBUG:_ids_to_rows] passages_df passage_id unique?:", self.passages_df["passage_id"].is_unique)
                print("[DEBUG:_ids_to_rows] duplicate passage_id count:", int(self.passages_df["passage_id"].duplicated().sum()))

            merged = ordered.merge(
                self.passages_df,
                on="passage_id",
                how="left",
                sort=False,
            )
            
            if len(merged) != len(ordered):
                raise ValueError(
                    "Mapping expanded (passage_id not unique or wrong passages/id_map loaded). "
                    f"ordered={len(ordered)} merged={len(merged)}"
                    )


            if self.debug:
                print("[DEBUG:_ids_to_rows] len(ordered):", len(ordered), "len(merged):", len(merged))
                if len(merged) != len(ordered):
                    # This is the smoking gun: one-to-many merge expansion.
                    vc = merged["passage_id"].value_counts(dropna=False).head(5)
                    print("[DEBUG:_ids_to_rows] TOP expanded passage_ids (value_counts):\n", vc)

            return merged

        # Fallback
        return self.passages_df.iloc[ids].copy()

    def timed_search(
        self,
        query: str,
        k: int = 10,
        efsearch: Optional[int] = None,
    ) -> Tuple[pd.DataFrame, Dict[str, float]]:

        if efsearch is not None:
            self.set_ef_search(efsearch)

        # 1) Encode query
        q_emb, t_embed_q = self._encode_query(query)

        # 2) FAISS search
        t0 = time.perf_counter()
        scores, idx = self.index.search(q_emb, int(k))
        t_search = (time.perf_counter() - t0) * 1000.0

        # Flatten
        idx = np.asarray(idx).reshape(-1)
        scores = np.asarray(scores).reshape(-1)

        if self.debug:
            print("\n[DEBUG:timed_search] k:", k)
            print("[DEBUG:timed_search] raw idx shape:", idx.shape, "raw scores shape:", scores.shape)
            print("[DEBUG:timed_search] raw idx[:10]:", idx[:10].tolist())

        # Filter invalid ids
        valid = idx != -1
        idx = idx[valid]
        scores = scores[valid]

        ids = idx.tolist()
        scr = scores.tolist()

        if self.debug:
            print("[DEBUG:timed_search] after valid filter len(ids):", len(ids), "len(scr):", len(scr))
            print("[DEBUG:timed_search] unique ids:", len(set(ids)))

        # 3) Map ids -> rows
        rows = self._ids_to_rows(ids)

        if self.debug:
            print("[DEBUG:timed_search] after _ids_to_rows len(rows):", len(rows), "len(scr):", len(scr))
            
            # if len(rows) > len(scr) then merge expanded rows OR something duplicated
            if len(rows) != len(scr):
                print("[DEBUG:timed_search] LENGTH MISMATCH detected right after mapping!")

        # If mapping failed for some rows, drop them and drop matching scores
        if "text" in rows.columns:
            if self.debug:
                print("[DEBUG:timed_search] before text notna filter len(rows):", len(rows), "len(scr):", len(scr))
                print("[DEBUG:timed_search] rows['text'] nulls:", int(rows["text"].isna().sum()))

            ok = rows["text"].notna().to_numpy()
            rows = rows.loc[ok].copy()
            # NOTE: this is where your crash happens if ok length != len(scr)
            scr = list(np.asarray(scr)[ok])

            if self.debug:
                print("[DEBUG:timed_search] after text notna filter len(rows):", len(rows), "len(scr):", len(scr))

        # Final safety alignment
        n = min(len(rows), len(scr))
        rows = rows.iloc[:n].copy()
        scr = scr[:n]

        rows["score"] = scr
        rows["rank"] = np.arange(1, n + 1)

        for col in ["doc_id", "text"]:
            if col not in rows.columns:
                rows[col] = None

        df = rows[["rank", "score", "passage_id", "doc_id", "text"]]

        timers = {
            "t_embed_q": float(t_embed_q),
            "t_search": float(t_search),
            "t_total_retrieval": float(t_embed_q + t_search),
        }
        return df, timers
```

```python
from typing import Optional, Tuple, Dict, List
import numpy as np
import pandas as pd
import time

from sklearn.feature_extraction.text import TfidfVectorizer


class HybridRetriever:
    """
    Hybrid retriever = dense (FAISS) + lexical (TF-IDF).

    - Uses an existing DenseRetriever instance.
    - Builds a TF-IDF index over passages_df[text_col].
    - At query time, gets candidates from both,
      min-max normalizes scores, and linearly combines them.

    alpha = weight for dense scores (0..1).
            (1-alpha) is weight for lexical scores.
    """

    def __init__(
        self,
        dense_retriever,               # DenseRetriever
        passages_df: pd.DataFrame,
        alpha: float = 0.5,
        text_col: str = "text",
        tfidf_stop_words: str = "english",
        tfidf_max_df: float = 0.95,
    ):
        self.dense = dense_retriever
        self.alpha = float(alpha)
        if not (0.0 <= self.alpha <= 1.0):
            raise ValueError("alpha must be between 0 and 1")

        if "passage_id" not in passages_df.columns:
            raise KeyError("passages_df must contain 'passage_id'")
        if text_col not in passages_df.columns:
            raise KeyError(f"passages_df must contain '{text_col}'")

        # Keep copy (do NOT reorder unless you mean to)
        self.passages_df = passages_df.copy()
        self.passages_df["passage_id"] = self.passages_df["passage_id"].astype(str)

        # Ensure required output columns exist
        for col in ["doc_id", "text"]:
            if col not in self.passages_df.columns:
                self.passages_df[col] = None

        # Map passage_id -> row index (fast lookups)
        # If passage_id has duplicates, last one wins (should not happen in your canonical pipeline).
        self.pid2row = {pid: i for i, pid in enumerate(self.passages_df["passage_id"].tolist())}

        # ---- Build TF-IDF index over passage text ----
        texts = self.passages_df[text_col].fillna("").astype(str).tolist()
        self.vectorizer = TfidfVectorizer(stop_words=tfidf_stop_words, max_df=tfidf_max_df)
        self.tfidf = self.vectorizer.fit_transform(texts)  # shape: [N, V]

    def set_ef_search(self, ef_search: int):
        """Only dense FAISS uses efSearch."""
        self.dense.set_ef_search(ef_search)

    @staticmethod
    def _minmax(x: np.ndarray) -> np.ndarray:
        if x.size == 0:
            return x
        xmin = float(x.min())
        xmax = float(x.max())
        if xmax <= xmin:
            return np.zeros_like(x, dtype=np.float32)
        return ((x - xmin) / (xmax - xmin)).astype(np.float32)

    def timed_search(
        self,
        query: str,
        k: int = 10,
        efsearch: Optional[int] = None,
        candidate_mult: int = 4,
    ) -> Tuple[pd.DataFrame, Dict[str, float]]:
        """
        Return a DataFrame with columns:
          [rank, score, passage_id, doc_id, text]
        where score is the HYBRID (dense + lexical) score.

        Timers dict has:
          t_embed_q, t_search, t_total_retrieval (ms)
        """
        t0 = time.perf_counter()

        # ---------- 1) DENSE SIDE ----------
        dense_k = max(k * candidate_mult, k)
        dense_df, dense_timers = self.dense.timed_search(
            query,
            k=dense_k,
            efsearch=efsearch,
        )

        # Dense results might be empty if index/search fails
        dense_pids = dense_df["passage_id"].astype(str).tolist() if len(dense_df) else []
        dense_scores_list = dense_df["score"].tolist() if len(dense_df) else []
        dense_score_map = dict(zip(dense_pids, dense_scores_list))

        # ---------- 2) LEXICAL SIDE (TF-IDF) ----------
        t_lex_start = time.perf_counter()

        q_vec = self.vectorizer.transform([query])          # [1, V]
        scores_lex = (self.tfidf @ q_vec.T).toarray().ravel()  # [N]

        lex_k = min(max(k * candidate_mult, k), scores_lex.shape[0])
        if lex_k > 0:
            lex_idx = np.argpartition(-scores_lex, lex_k - 1)[:lex_k]
        else:
            lex_idx = np.array([], dtype=int)

        lex_pids = self.passages_df.iloc[lex_idx]["passage_id"].astype(str).tolist() if lex_k > 0 else []

        t_lex_end = time.perf_counter()

        # Map lexical scores
        lex_score_map = {pid: float(scores_lex[self.pid2row[pid]]) for pid in lex_pids}

        # ---------- 3) FUSION ----------
        # Union of dense + lexical candidates (preserve order)
        candidate_pids = list(dict.fromkeys(dense_pids + lex_pids))

        if not candidate_pids:
            # return empty result with timers
            empty = pd.DataFrame(columns=["rank", "score", "passage_id", "doc_id", "text"])
            t_total_ms = (time.perf_counter() - t0) * 1000.0
            timers = {
                "t_embed_q": float(dense_timers.get("t_embed_q", 0.0)) if isinstance(dense_timers, dict) else 0.0,
                "t_search": max(t_total_ms, 0.0),
                "t_total_retrieval": t_total_ms,
            }
            return empty, timers

        dense_scores = np.array([dense_score_map.get(pid, 0.0) for pid in candidate_pids], dtype=np.float32)
        lex_scores = np.array([lex_score_map.get(pid, 0.0) for pid in candidate_pids], dtype=np.float32)

        dense_norm = self._minmax(dense_scores)
        lex_norm = self._minmax(lex_scores)

        hybrid_scores = self.alpha * dense_norm + (1.0 - self.alpha) * lex_norm

        # Top-k
        k_eff = min(int(k), len(candidate_pids))
        order = np.argsort(-hybrid_scores)[:k_eff]
        final_pids = [candidate_pids[i] for i in order]
        final_scores = hybrid_scores[order]

        # Lookup rows
        final_idx = [self.pid2row[pid] for pid in final_pids]
        rows = self.passages_df.iloc[final_idx].copy()
        rows["score"] = final_scores.astype(np.float32)
        rows["rank"] = np.arange(1, len(rows) + 1)

        df = rows[["rank", "score", "passage_id", "doc_id", "text"]]

        # ---------- Timers ----------
        t_total_ms = (time.perf_counter() - t0) * 1000.0
        t_embed_q = float(dense_timers.get("t_embed_q", 0.0))
        t_search = max(t_total_ms - t_embed_q, 0.0)

        timers = {
            "t_embed_q": t_embed_q,
            "t_search": t_search,
            "t_total_retrieval": t_total_ms,
            "t_lexical_ms": (t_lex_end - t_lex_start) * 1000.0,  # extra (optional)
        }

        return df, timers
```

```python
from typing import Optional, Tuple, Dict
import numpy as np
import pandas as pd
import time

from sklearn.feature_extraction.text import CountVectorizer


class BM25Retriever:
    """
    Pure lexical BM25 retriever over passages_df['text'].

    - Uses CountVectorizer to build a term-count matrix.
    - Precomputes BM25 stats (idf, doc lengths, avgdl).
    - Exposes the SAME interface as DenseRetriever / HybridRetriever:
        timed_search(query, k, efsearch=None) -> (df, timers)
    """

    def __init__(
        self,
        passages_df: pd.DataFrame,
        k1: float = 1.5,
        b: float = 0.75,
        text_col: str = "text",
    ):
        if "passage_id" not in passages_df.columns:
            raise KeyError("passages_df must contain 'passage_id'")
        if text_col not in passages_df.columns:
            raise KeyError(f"passages_df must contain '{text_col}'")

        self.passages_df = passages_df.copy()

        # Ensure required output columns exist (so evaluation doesn't crash)
        self.passages_df["passage_id"] = self.passages_df["passage_id"].astype(str)
        if "doc_id" not in self.passages_df.columns:
            self.passages_df["doc_id"] = None
        if "text" not in self.passages_df.columns:
            self.passages_df["text"] = self.passages_df[text_col]

        self.k1 = float(k1)
        self.b = float(b)
        self.text_col = text_col

        # 1) Build term-count matrix
        texts = self.passages_df[self.text_col].fillna("").astype(str).tolist()
        self.vectorizer = CountVectorizer()
        X = self.vectorizer.fit_transform(texts)  # sparse (n_docs, n_terms)
        self.X = X

        # 2) Precompute BM25 stats
        self.N = int(X.shape[0])

        # document frequencies (df) per term
        df_t = (X > 0).sum(axis=0).A1.astype(np.float32)  # (n_terms,)
        # BM25 idf
        self.idf = np.log(1.0 + (self.N - df_t + 0.5) / (df_t + 0.5)).astype(np.float32)

        # document lengths and average length
        self.doc_len = X.sum(axis=1).A1.astype(np.float32)  # (n_docs,)
        self.avgdl = float(np.mean(self.doc_len)) if self.N > 0 else 0.0

    def _score(self, query: str) -> np.ndarray:
        """
        Compute BM25 scores for all documents for a single query.
        Returns float32 array of shape (n_docs,).
        """
        if self.N == 0:
            return np.zeros(0, dtype=np.float32)

        q_vec = self.vectorizer.transform([query])  # (1, n_terms)
        q_indices = q_vec.indices  # non-zero term indices in query

        if q_indices.size == 0:
            return np.zeros(self.N, dtype=np.float32)

        # Sparse slice: (n_docs, n_query_terms)
        X_sub = self.X[:, q_indices]

        # Convert only this slice to dense (ok for 10k passages; safer than densifying full X)
        tf = X_sub.toarray().astype(np.float32)  # (n_docs, n_query_terms)

        dl = self.doc_len[:, None]  # (n_docs, 1)
        idf = self.idf[q_indices][None, :]  # (1, n_query_terms)

        k1 = self.k1
        b = self.b
        avgdl = self.avgdl if self.avgdl > 0 else 1.0

        denom = tf + k1 * (1.0 - b + b * dl / avgdl)
        denom = np.maximum(denom, 1e-9)

        num = tf * (k1 + 1.0)
        scores = (idf * (num / denom)).sum(axis=1)  # (n_docs,)

        return scores.astype(np.float32)

    def timed_search(
        self,
        query: str,
        k: int = 10,
        efsearch: Optional[int] = None,  # ignored (dense-only)
    ) -> Tuple[pd.DataFrame, Dict[str, float]]:
        """
        Same interface as DenseRetriever / HybridRetriever:
          - Ignores efsearch.
          - Returns hits dataframe with [rank, score, passage_id, doc_id, text].
          - Returns timers dict: t_embed_q, t_search, t_total_retrieval (ms).
        """
        t0 = time.perf_counter()

        scores = self._score(query)  # (n_docs,)
        n_docs = scores.shape[0]

        k_eff = min(int(k), n_docs)
        if k_eff <= 0:
            empty_df = pd.DataFrame(columns=["rank", "score", "passage_id", "doc_id", "text"])
            t_total_ms = (time.perf_counter() - t0) * 1000.0
            timers = {"t_embed_q": 0.0, "t_search": t_total_ms, "t_total_retrieval": t_total_ms}
            return empty_df, timers

        # Top-k indices
        topk_idx = np.argpartition(-scores, k_eff - 1)[:k_eff]
        topk_idx = topk_idx[np.argsort(-scores[topk_idx])]

        rows = self.passages_df.iloc[topk_idx].copy()
        rows["score"] = scores[topk_idx]
        rows["rank"] = np.arange(1, len(rows) + 1)

        df = rows[["rank", "score", "passage_id", "doc_id", "text"]]

        t_total_ms = (time.perf_counter() - t0) * 1000.0
        timers = {"t_embed_q": 0.0, "t_search": t_total_ms, "t_total_retrieval": t_total_ms}
        return df, timers
```

```python
import re
from typing import List, Optional, Tuple, Dict

import numpy as np
import pandas as pd
import tracemalloc


def _normalize_text(s: str) -> str:
    s = str(s)  # handle None/NaN
    s = s.lower().strip()
    s = re.sub(r"\s+", " ", s)
    return s


def _has_answer(passage: str, answers: List[str]) -> bool:
    # passage can be NaN/None if mapping failed; normalize safely
    if passage is None or (isinstance(passage, float) and np.isnan(passage)):
        return False
    p = _normalize_text(passage)
    for a in answers:
        if a is None or (isinstance(a, float) and np.isnan(a)):
            continue
        if _normalize_text(a) in p:
            return True
    return False


def evaluate_retriever(
    retriever,
    queries_df: pd.DataFrame,
    q_col: str,
    supervision: str,                   # "gold_id" or "answers"
    gold_col: Optional[str] = None,     # when supervision="gold_id"
    answers_col: Optional[str] = None,  # when supervision="answers"
    k_list: List[int] = [1, 3, 5, 10],
    efsearch: Optional[int] = None
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Runs retrieval over queries_df with a given retriever and returns:
      - per_query_df: one row per query with hits@k, latency, memory, and top-k info
      - summary_df:   aggregate Recall@k, MRR@k, and per-K latency/memory metrics

    retriever must implement:
      timed_search(query, k, efsearch=None) -> (df, timers)
    """
    assert supervision in {"gold_id", "answers"}

    if supervision == "gold_id" and not gold_col:
        raise ValueError("gold_col must be provided when supervision='gold_id'")
    if supervision == "answers" and not answers_col:
        raise ValueError("answers_col must be provided when supervision='answers'")

    # Optionally set efSearch once (if retriever supports it)
    if efsearch is not None and hasattr(retriever, "set_ef_search"):
        retriever.set_ef_search(efsearch)

    per_rows = []
    rr_lists = {f"MRR@{k}": [] for k in k_list}
    recall_counts = {f"Recall@{k}": 0 for k in k_list}

    tracemalloc.start()

    max_k = int(max(k_list))
    print("efsearch = ", efsearch)

    for i, row in queries_df.iterrows():
        question = row[q_col]

        tracemalloc.reset_peak()

        # --- 1) Run retrieval ---
        df, timers = retriever.timed_search(question, k=max_k, efsearch=efsearch)

        # Ensure df has expected columns even if empty
        if df is None or len(df) == 0:
            ranked_pids = []
            ranked_texts = []
            ranked_scores = []
        else:
            # Normalize types
            ranked_pids = df.get("passage_id", pd.Series([], dtype=str)).astype(str).tolist()
            ranked_texts = df.get("text", pd.Series([], dtype=object)).tolist()
            ranked_scores = df.get("score", pd.Series([], dtype=float)).tolist()

        # --- 2) Peak memory (MB) ---
        _, peak_bytes = tracemalloc.get_traced_memory()
        mem_peak_mb = peak_bytes / (1024 * 1024)

        # --- 3) Latency (ms) ---
        retrieval_latency_ms = timers.get("t_total_retrieval", timers.get("t_search", np.nan))

        # --- 4) Compute relevance ---
        if supervision == "gold_id":
            gold_pid = str(row[gold_col])

            for k in k_list:
                recall_counts[f"Recall@{k}"] += int(gold_pid in ranked_pids[:k])

            rr = 0.0
            found_rank = None
            for r_idx, pid in enumerate(ranked_pids, start=1):
                if pid == gold_pid:
                    rr = 1.0 / r_idx
                    found_rank = r_idx
                    break

            for k in k_list:
                rr_lists[f"MRR@{k}"].append(rr if (found_rank is not None and found_rank <= k) else 0.0)

            hit_flags = {f"hit@{k}": int(gold_pid in ranked_pids[:k]) for k in k_list}

        else:  # supervision == "answers"
            answers = row[answers_col]

            # Normalize answers into a list[str]
            if answers is None or (isinstance(answers, float) and np.isnan(answers)):
                answers = []
            elif isinstance(answers, str):
                answers = [answers]
            elif not isinstance(answers, (list, tuple, np.ndarray)):
                answers = [str(answers)]
            else:
                answers = list(answers)

            for k in k_list:
                topk_texts = ranked_texts[:k]
                hit = any(_has_answer(t, answers) for t in topk_texts)
                recall_counts[f"Recall@{k}"] += int(hit)

            rr = 0.0
            found_rank = None
            for r_idx, ttxt in enumerate(ranked_texts, start=1):
                if _has_answer(ttxt, answers):
                    rr = 1.0 / r_idx
                    found_rank = r_idx
                    break

            for k in k_list:
                rr_lists[f"MRR@{k}"].append(rr if (found_rank is not None and found_rank <= k) else 0.0)

            hit_flags = {f"hit@{k}": int(any(_has_answer(t, answers) for t in ranked_texts[:k])) for k in k_list}

        # --- 5) Store per-query record ---
        per_rows.append({
            "q_idx": i,
            "question": question,
            **hit_flags,
            "retrieval_latency_ms": retrieval_latency_ms,
            "mem_peak_mb": mem_peak_mb,
            "topk_pids": ranked_pids,
            "topk_scores": ranked_scores,
            "topk_texts": ranked_texts,
        })

    tracemalloc.stop()

    per_query_df = pd.DataFrame(per_rows)
    n = len(per_query_df)

    # --- 6) Aggregate metrics per K ---
    summary: Dict[str, object] = {}

    for k in k_list:
        summary[f"Recall@{k}"] = recall_counts[f"Recall@{k}"] / n if n > 0 else 0.0
        summary[f"MRR@{k}"] = float(np.mean(rr_lists[f"MRR@{k}"])) if rr_lists[f"MRR@{k}"] else 0.0

        # Latency/memory across ALL queries (recommended)
        summary[f"mean_retrieval_latency_ms@{k}"] = float(np.nanmean(per_query_df["retrieval_latency_ms"])) if n > 0 else np.nan
        summary[f"mean_mem_peak_mb@{k}"] = float(np.nanmean(per_query_df["mem_peak_mb"])) if n > 0 else np.nan
        summary[f"max_mem_peak_mb@{k}"] = float(np.nanmax(per_query_df["mem_peak_mb"])) if n > 0 else np.nan

        # (Optional) If you still want hit-only stats, keep these extra fields:
        hit_mask = per_query_df[f"hit@{k}"] == 1
        summary[f"mean_latency_on_hits_ms@{k}"] = float(np.nanmean(per_query_df.loc[hit_mask, "retrieval_latency_ms"])) if hit_mask.any() else np.nan

    summary_df = pd.DataFrame([summary])
    return per_query_df, summary_df
```

```python
def _count_tokens_with_model(llm, text: str) -> int:
    return len(llm.tokenize(text.encode("utf-8")))

def _truncate_to_tokens(llm, text: str, max_tokens: int) -> str:
    toks = llm.tokenize(text.encode("utf-8"))
    toks = toks[:max_tokens]
    return llm.detokenize(toks).decode("utf-8", errors="ignore")

def build_rag_prompt_budgeted(question, contexts, llm, max_input_tokens=512):
    system_part = (
        "You are a question answering assistant. "
        "Use ONLY the given context. "
        "Return ONLY the short answer phrase copied verbatim from the context. "
        "Do not add any explanation or extra words. "
        "If the answer is not in the context, say \"I don't know\"."
    )

    sep = "\n\n---\n\n"

    base_prompt = (
        system_part
        + "\n\nContext:\n\n"
        + f"Question: {question}\n"
        + "Answer:"
    )

    base_len = _count_tokens_with_model(llm, base_prompt)
    context_budget = max_input_tokens - base_len
    if context_budget <= 0:
        return base_prompt

    included = []
    used = 0

    for ctx in contexts:
        # include separator cost if this isn't the first chunk
        sep_cost = _count_tokens_with_model(llm, sep) if included else 0
        ctx_cost = _count_tokens_with_model(llm, ctx)

        if used + sep_cost + ctx_cost <= context_budget:
            included.append(ctx)
            used += sep_cost + ctx_cost
        else:
            # Try truncating this chunk to fit remaining budget
            remaining = context_budget - used - sep_cost
            if remaining > 0:
                truncated = _truncate_to_tokens(llm, ctx, remaining)
                if truncated.strip():
                    included.append(truncated)
                    used += sep_cost + _count_tokens_with_model(llm, truncated)
            # don't break; try next contexts
            continue

    final_prompt = (
        system_part
        + "\n\nContext:\n\n"
        + (sep.join(included) if included else "")
        + "\n\nQuestion: "
        + question
        + "\nAnswer:"
    )
    return final_prompt
```

```python
import re

def clean_answer(raw_answer: str) -> str:
    """
    Post-process the raw model output to enforce a short phrase answer.
    Strategy:
      - Take the first non-empty line.
      - Strip trailing '.', '?', '!' if present.
      - Strip common prefixes ("The answer is", "Answer:", "Assistant:", etc.).
      - Strip surrounding quotes.
      - If the cleaned answer starts with 'Context', drop that word.
    """
    if not raw_answer:
        return raw_answer

    # 1) First non-empty line
    lines = [l.strip() for l in raw_answer.splitlines() if l.strip()]
    if not lines:
        return ""

    line = lines[0]
    ans = line.strip()

    # 2) Drop only trailing sentence-ending punctuation (keep the rest of the phrase)
    ans = ans.rstrip(".?!").strip()

    # 3) Remove common leading phrases (case-insensitive)
    lower = ans.lower()
    prefixes = [
        "the answer is",
        "the answer:",
        "answer is",
        "answer:",
        "it is",
        "it's",
        "it was",
        "the country is",
        "assistant:",
        "model:",
    ]
    for p in prefixes:
        if lower.startswith(p):
            ans = ans[len(p):].strip(" ,:-")
            break

    # 4) Strip surrounding quotes if present
    ans = ans.strip()
    if len(ans) >= 2 and (
        (ans[0] == ans[-1] == '"') or
        (ans[0] == ans[-1] == "'")
    ):
        ans = ans[1:-1].strip()

    # 5) If it starts with 'context', drop that token
    if ans.lower().startswith("context"):
        parts = ans.split()
        ans = " ".join(parts[1:]) if len(parts) > 1 else ""

    # 6) Final trim + collapse multiple spaces
    ans = re.sub(r"\s+", " ", ans).strip()

    return ans
```

```python
def generate_answer_from_context(
    question: str,
    contexts: list[str],
    max_input_tokens: int = 512,
    max_new_tokens: int = 32,
) -> str:
   
    # 1) Build prompt with context budgeting (uses qwen_llm.tokenize internally)
    prompt = build_rag_prompt_budgeted(question, contexts, llm=qwen_llm, max_input_tokens=max_input_tokens)

    # 2) Call the llama-cpp model directly
    # n_ctx in qwen_llm must be >= max_input_tokens + max_new_tokens
    result = qwen_llm(
        prompt,
        max_tokens=max_new_tokens,   # number of new tokens to generate
        temperature=0.0,             # fully greedy / deterministic
        top_p=1.0,
    )

    # 3) Extract text and clean
    raw_answer = result["choices"][0]["text"].strip()
    return clean_answer(raw_answer)
```

```python
import pandas as pd
import pyarrow.parquet as pq

RAG_OUTPUT_PATH_HYBRID = PROJECT_ROOT / "results" / "rag_answers_hybrid.parquet"

# You already discovered these:
good_cols = [
    "question_id",
    "question",
    "generated_answer",
    "gold_answer_text",
]

pf = pq.ParquetFile(RAG_OUTPUT_PATH_HYBRID)

data = {name: [] for name in good_cols}

for rg_index in range(pf.num_row_groups):
    rg_table = pf.read_row_group(rg_index, columns=good_cols)
    rg_table = rg_table.combine_chunks()
    for name in good_cols:
        col = rg_table[name]
        data[name].extend(col.to_pylist())

rag_results_df_hybrid = pd.DataFrame(data)

print("Columns in hybrid df:", rag_results_df_hybrid.columns.tolist())
rag_results_df_hybrid.head(1)
```

```python
import re
import string
import numpy as np

def normalize_text(s: str) -> str:
    """
    Lowercase, remove punctuation, articles, and extra whitespace.
    This matches common EM/F1 implementations used in QA benchmarks.
    """
    if s is None:
        return ""

    # Lowercase
    s = s.lower()

    # Remove punctuation
    exclude = set(string.punctuation)
    s = "".join(ch for ch in s if ch not in exclude)

    # Remove articles
    s = re.sub(r"\b(a|an|the)\b", " ", s)

    # Normalize whitespace
    s = " ".join(s.split())

    return s
```

```python
def exact_match_score(pred: str, gold: str) -> float:
    """
    1.0 if normalized prediction == normalized gold, else 0.0.
    """
    return float(normalize_text(pred) == normalize_text(gold))


def f1_score_qa(pred: str, gold: str) -> float:
    """
    Token-level F1 between prediction and a single gold answer.
    """
    pred_tokens = normalize_text(pred).split()
    gold_tokens = normalize_text(gold).split()

    if len(pred_tokens) == 0 and len(gold_tokens) == 0:
        return 1.0
    if len(pred_tokens) == 0 or len(gold_tokens) == 0:
        return 0.0

    common = {}
    for t in pred_tokens:
        if t in gold_tokens:
            common[t] = common.get(t, 0) + 1

    num_common = 0
    for t in common:
        num_common += min(pred_tokens.count(t), gold_tokens.count(t))

    if num_common == 0:
        return 0.0

    precision = num_common / len(pred_tokens)
    recall    = num_common / len(gold_tokens)
    f1        = 2 * precision * recall / (precision + recall)
    return f1
```

```python
def get_gold_answers(row) -> list[str]:
    """
    Turn gold_answer_text into a list of gold answer strings.
    Supports:
      - already a list
      - a string with answers separated by " , "
      - missing/NaN → empty list
    """
    val = row.get("gold_answer_text", None)

    # If already a list
    if isinstance(val, list):
        return [str(a).strip() for a in val if str(a).strip()]

    # If NaN / missing
    if val is None or (isinstance(val, float) and np.isnan(val)):
        return []

    # Otherwise treat as a string with " , " delimiter
    s = str(val)
    parts = [p.strip() for p in s.split(" , ") if p.strip()]
    return parts


def best_em_f1_for_row(pred: str, golds: list[str]) -> tuple[float, float]:
    """
    For one prediction and a list of gold answers, return:
      (best_exact_match, best_f1)
    """
    if not golds:
        return np.nan, np.nan  # skip in aggregate

    em_scores = [exact_match_score(pred, g) for g in golds]
    f1_scores = [f1_score_qa(pred, g) for g in golds]

    return max(em_scores), max(f1_scores)


def clean_pred(pred: str) -> str:
    if pred is None:
        return ""
    pred = pred.strip()

    pred = re.sub(r"^(the answer is|answer is|answer:)\s*", "", pred, flags=re.I)
    pred = pred.strip(" \"'.,;:")

    return pred


def add_em_f1_metrics(df: pd.DataFrame) -> pd.DataFrame:
    """
    Compute EM and F1 for each row and add them as 'em' and 'f1' columns.
    Modifies df in-place and also returns it.
    """
    em_list = []
    f1_list = []

    for _, row in df.iterrows():
        pred = clean_pred(row["generated_answer"])
        gold_answers = get_gold_answers(row)

        em, f1 = best_em_f1_for_row(pred, gold_answers)
        em_list.append(em)
        f1_list.append(f1)

    df["em"] = em_list
    df["f1"] = f1_list
    return df
```

```python
# Drop rows with NaN EM/F1 (questions that had no gold answers)
rag_results_df_bm25 = add_em_f1_metrics(rag_results_df_bm25)

# keep only rows where EM/F1 are defined
valid_df = rag_results_df_bm25.dropna(subset=["em", "f1"])

avg_em = valid_df["em"].mean()
avg_f1 = valid_df["f1"].mean()

print("BM25 Results are as follows:")
print(f"Number of evaluated questions: {len(valid_df)}")
print(f"Exact Match (EM):                 {avg_em:.4f}")
print(f"F1:                               {avg_f1:.4f}")



```
