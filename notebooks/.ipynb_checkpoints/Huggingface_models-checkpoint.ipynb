```python
from huggingface_hub import login, snapshot_download
import sys
import os

# 1) Login (once per machine / env)
login()  # or login(token="hf_...")

# 2) Download Phi-3 GGUF
MODEL_DIR = os.path.join(sys.prefix, "models", "QuantFactory/Phi-3-mini-4k-instruct-GGUF")
snapshot_download(
    repo_id="QuantFactory/Phi-3-mini-4k-instruct-GGUF",  # adjust if needed
    local_dir=MODEL_DIR,
    local_dir_use_symlinks=False,
    allow_patterns="*Q4_K_M.gguf",
)

# 3) Download Qwen2.5-3B GGUF
MODEL_DIR = os.path.join(sys.prefix, "models", "Qwen/Qwen2.5-3B-Instruct-GGUF")
snapshot_download(
    repo_id="Qwen/Qwen2.5-3B-Instruct-GGUF",  # adjust if needed
    local_dir=MODEL_DIR,
    local_dir_use_symlinks=False,
    allow_patterns="*Q4_K_M.gguf",
)
```
